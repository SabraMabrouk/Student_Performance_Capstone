## Imports

```python
import pandas as pd
import numpy as np
import os
import pickle
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import scale
from sklearn.model_selection import train_test_split, cross_validate, GridSearchCV, learning_curve
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error as mse, mean_absolute_error as mae
from sklearn.pipeline import make_pipeline,Pipeline
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_regression
from xgboost import XGBRegressor
```

```python
df_student = pd.read_csv('./data/data_student_cleaned.csv')
```

# Modeling

## Regression
As we expected the G1 and G2 grades to have a significant impact on G3, three configurations were tested for each modele
- A: Consider all features except G3 (Target)
- B: Similar to A and drop G2 too
- C: Similar to B and drop G1 too

```python

data = {
    'configA' : [df_student_numeric.loc[df_student_numeric['course_math']==1].drop(columns=['G3','course_math']),df_student_numeric.loc[df_student_numeric['course_math']==0].drop(columns=['G3','course_math'])],
    'configB' : [df_student_numeric.loc[df_student_numeric['course_math']==1].drop(columns=['G2','G3','course_math']),df_student_numeric.loc[df_student_numeric['course_math']==0].drop(columns=['G2','G3','course_math'])],
    'configC' : [df_student_numeric.loc[df_student_numeric['course_math']==1].drop(columns=['G1','G2','G3','course_math']),df_student_numeric.loc[df_student_numeric['course_math']==0].drop(columns=['G1','G2','G3','course_math'])]
}
```

```python
lang=['math','portuguese']
Models=[LinearRegression(),RandomForestRegressor(random_state=47),XGBRegressor(objective= 'reg:squarederror')]
#metrics={'configA':{'math':[],'portuguese':[]},'configB':{'math':[],'portuguese':[]},'configC':{'math':[],'portuguese':[]}}
cross_valid={'configA':{'math':[],'portuguese':[]},'configB':{'math':[],'portuguese':[]},'configC':{'math':[],'portuguese':[]}}
best_param={'configA':{'math':[],'portuguese':[]},'configB':{'math':[],'portuguese':[]},'configC':{'math':[],'portuguese':[]}}
features_imp={'configA':{'math':[],'portuguese':[]},'configB':{'math':[],'portuguese':[]},'configC':{'math':[],'portuguese':[]}}

target = [df_student_numeric.loc[df_student_numeric['course_math']==1,'G3'],df_student_numeric.loc[df_student_numeric['course_math']==0,'G3']]

for config,dfs in data.items():
    for ind,df in enumerate(dfs):
        
        # Train/test split
        X_train, X_test, y_train, y_test = train_test_split(df, target[ind], test_size=0.3, random_state=47)
        for mod in Models:
            
            # Define the pipeline
            if str(mod) =='LinearRegression()':
                pipe = make_pipeline( 
                    StandardScaler(),
                    SelectKBest(f_regression),
                    LinearRegression()
                )
                pipe.get_params().keys()
                k = [k+1 for k in range(len(X_train.columns))]
                grid_params = {
                    'selectkbest__k': k,
                    'standardscaler': [StandardScaler(), MinMaxScaler(), None]
                }
            elif str(mod) =='RandomForestRegressor(random_state=47)':
                pipe = make_pipeline(
                        StandardScaler(),
                        RandomForestRegressor(random_state=47)
                )
                n_est = [int(n) for n in np.logspace(start=1, stop=3, num=20)]
                grid_params = {
                    'randomforestregressor__n_estimators': n_est,
                    'standardscaler': [StandardScaler(), MinMaxScaler(), None],
                    }
            else:
                pipe = Pipeline(steps=[("scaler", StandardScaler()),
                                       ("model", XGBRegressor(objective= 'reg:squarederror'))])
                grid_params = {
                    'model__learning_rate': np.arange(0.05, 1, 0.05),
                    'model__max_depth': np.arange(3, 10, 1),
                    'model__n_estimators': np.arange(50, 200, 50),
                    #'scaler__standardscaler': [StandardScaler(), MinMaxScaler(), None]
                }
        
            grid_cv = GridSearchCV(pipe, param_grid=grid_params, cv=5, n_jobs=-1)
            grid_cv.fit(X_train, y_train)
            best_param[config][lang[ind]].append(grid_cv.best_params_)
            if str(mod) =='LinearRegression()':
                selected = grid_cv.best_estimator_.named_steps.selectkbest.get_support()
                coefs = grid_cv.best_estimator_.named_steps.linearregression.coef_
                features = X_train.columns[selected]
                features_imp[config][lang[ind]].append([pd.Series(coefs, index=features).sort_values(ascending=False)])
            elif str(mod) =='RandomForestRegressor(random_state=47)':
                imps = grid_cv.best_estimator_.named_steps.randomforestregressor.feature_importances_
                features_imp[config][lang[ind]].append([pd.Series(imps, index=X_train.columns).sort_values(ascending=False)])

            else:
                imps = grid_cv.best_estimator_.named_steps.model.feature_importances_
                features_imp[config][lang[ind]].append([pd.Series(imps, index=X_train.columns).sort_values(ascending=False)])
               
            neg_mae = cross_validate(grid_cv.best_estimator_, X_train, y_train, 
                            scoring='neg_mean_absolute_error', cv=5, n_jobs=-1)
            cross_valid[config][lang[ind]].append([np.mean(-1 * neg_mae['test_score']), np.std(-1 * neg_mae['test_score'])])
            #y_tr_pred = grid_cv.predict(X_train)
            #y_te_pred = grid_cv.predict(X_test)
            #print('r2_train = ',r2_score(y_train, y_tr_pred),'r2_test = ', r2_score(y_test, y_te_pred))
            #print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
            #print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```

```python
for config in data.keys():
    print(config,'\n')
    print('Math results')
    for ind,i in enumerate(cross_valid[config]['math']):
        print('Model: ',str(Models[ind])[0:str(Models[ind]).index('(')])
        print('mae_mean',i[0],'mae_std',i[1])
    print('-------------------------------------------------------------------------------')    
    print('portuguese results')
    for ind,i in enumerate(cross_valid[config]['portuguese']):
        print('Model: ',str(Models[ind])[0:str(Models[ind]).index('(')])
        print('mae_mean',i[0],'mae_std',i[1])
    print('\n*******************************************************************************\n') 
```

```python
for config in data.keys():
    print(config,'\n')
    print('Math best parameters')
    for ind,i in enumerate(best_param[config]['math']):
        print('Model: ',str(Models[ind])[0:str(Models[ind]).index('(')])
        print('Best parameters',i)
    print('-------------------------------------------------------------------------------')    
    print('portuguese best parameters')
    for ind,i in enumerate(best_param[config]['portuguese']):
        print('Model: ',str(Models[ind])[0:str(Models[ind]).index('(')])
        print('Best parameters',i)
    print('\n*******************************************************************************\n') 
```

```python
for config in data.keys():
    print(config,'\n')
    print('Math best parameters')
    for ind,i in enumerate(features_imp[config]['math']):
        print('Model: ',str(Models[ind])[0:str(Models[ind]).index('(')])
        print('Best parameters',i)
    print('-------------------------------------------------------------------------------')    
    print('portuguese best parameters')
    for ind,i in enumerate(features_imp[config]['portuguese']):
        print('Model: ',str(Models[ind])[0:str(Models[ind]).index('(')])
        print('Best parameters',i)
    print('\n*******************************************************************************\n') 
```

## Classification
We consider two classification problems:
- Binary classification:
    - G3≥10: Pass
    - G3<10: fail
- 5-Level classification:
    - 16-20: Excellent
    - 14-15: Good
    - 12-13: Satisfactory
    - 10-11: Sufficient
    - 0-9: Fail

### Binary classification

```python
def binary_funct(x):
    if x >= 10:
        return 'pass'
    else:
        return 'fail'

df_student_numeric.loc[:,'binary'] = list(map(binary_funct,df_student.loc[:,'G3']))
```

## Configuration A

```python
data = df_student.drop(columns=['G3'])
data_numeric=df_student_numeric.drop(columns=['G3'])
```

### Train/test split

```python
X_train, X_test, y_train, y_test = train_test_split(data_numeric, 
                                                    df_student_numeric['G3'], test_size=0.3, 
                                                    random_state=47)
```

### Predictive Models

#### Linear regression model

```python
# scale the features
#scaler = StandardScaler()
#scaler.fit(X_train)
#X_tr_scaled = scaler.transform(X_train)
#X_te_scaled = scaler.transform(X_test)

# train a model
#lm = LinearRegression().fit(X_tr_scaled, y_train)
#y_tr_pred = lm.predict(X_tr_scaled)
#y_te_pred = lm.predict(X_te_scaled)

# Assess model performance
#print('r2_train = ',r2_score(y_train, y_tr_pred),'r2_test = ', r2_score(y_test, y_te_pred))
#print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
#print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```

***Pipeline creation and hyperparameter tuning*** 

```python
pipe = make_pipeline( 
    StandardScaler(),
    SelectKBest(f_regression),
    LinearRegression()
)

pipe.get_params().keys()
k = [k+1 for k in range(len(X_train.columns))]
grid_params = {
    'selectkbest__k': k,
    'standardscaler': [StandardScaler(), MinMaxScaler(), None]
}
lr_grid_cv = GridSearchCV(pipe, param_grid=grid_params, cv=5, n_jobs=-1)
lr_grid_cv.fit(X_train, y_train)
lr_grid_cv.best_params_
```

***Features importance***

```python
selected = lr_grid_cv.best_estimator_.named_steps.selectkbest.get_support()
coefs = lr_grid_cv.best_estimator_.named_steps.linearregression.coef_
features = X_train.columns[selected]
pd.Series(coefs, index=features).sort_values(ascending=False)
```

***Cross validation***

```python
lr_best_cv_results = cross_validate(lr_grid_cv.best_estimator_, X_train, y_train, cv=5)
lr_best_scores = lr_best_cv_results['test_score']
np.mean(lr_best_scores), np.std(lr_best_scores)
```

***Assess performance***

```python
y_tr_pred = lr_grid_cv.predict(X_train)
y_te_pred = lr_grid_cv.predict(X_test)

print('r2_train = ',r2_score(y_train, y_tr_pred),'r2_test = ', r2_score(y_test, y_te_pred))
print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```
