## Imports

```python
import pandas as pd
import numpy as np
import os
import pickle
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import scale
from sklearn.model_selection import train_test_split, cross_validate, GridSearchCV
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor,GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_squared_error as mse, mean_absolute_error as mae
from sklearn.pipeline import make_pipeline,Pipeline
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_regression
from xgboost import XGBRegressor
```

## Load The Data
We start by loading  the saved data from the previous step.

```python
df_student = pd.read_csv('./data/data_student_cleaned_pre.csv')
df_student.head()
```

Let's check the columns and their types. All features are numeric and there is no null value.

```python
df_student.info()
```

We separate the data for the math course and the portuguese course. Prediction models will be applied on these data seprately.

```python
df_math = df_student[df_student['course']==0]
df_port = df_student[df_student['course']==1]
```

# Train test split

We split the data into a training and test sets:
- 30% for test
- 70% for training

```python
# split the math data
X_math = df_math.drop(columns=['G3','course'])
y_math = df_math['G3']
X_train_math, X_test_math, y_train_math, y_test_math = train_test_split(X_math, y_math, test_size=0.3, random_state=47)

# split the portuguese data
X_port = df_port.drop(columns=['G3','course'])
y_port = df_port['G3']
X_train_port, X_test_port, y_train_port, y_test_port = train_test_split(X_port, y_port, test_size=0.3, random_state=47)
```

```python
print(X_train_math.shape)
print(X_test_math.shape)
```

```python
print(X_train_port.shape)
print(X_test_port.shape)
```

# Scale tha data

```python
## Sacle the math data
numeric_features = ['age','failures','absences']
ordinal_features = ['Medu', 'Fedu', 'traveltime', 'studytime', 'famrel','freetime', 'goout', 'Dalc', 'Walc', 'health']
X_train_rest_math = X_train_math.loc[:, ~X_train_math.columns.isin(ordinal_features+numeric_features)]
X_test_rest_math = X_test_math.loc[:, ~X_test_math.columns.isin(ordinal_features+numeric_features)]
# Initialize  the scalers
scaler_mm = MinMaxScaler()
scaler_std = StandardScaler()

# fit and transform
scaler_mm.fit(X_train_math[ordinal_features])
scaler_std.fit(X_train_math[numeric_features])
X_train_scaled_num_math =  pd.DataFrame(scaler_std.transform(X_train_math[numeric_features]), columns=numeric_features)  #scaler_std.transform(X_train[numeric_features])
X_train_scaled_ord_math = pd.DataFrame(scaler_mm.transform(X_train_math[ordinal_features]), columns=ordinal_features)#scaler_mm.transform(X_train[ordinal_features])
X_train_scaled_math = pd.concat([X_train_scaled_num_math.reset_index(drop=True),X_train_scaled_ord_math.reset_index(drop=True),X_train_rest_math.reset_index(drop=True)],axis=1)


# transform the test data
X_test_scaled_num_math = pd.DataFrame(scaler_std.transform(X_test_math[numeric_features]), columns=numeric_features) #scaler_std.transform(X_test[numeric_features])
X_test_scaled_ord_math = pd.DataFrame(scaler_mm.transform(X_test_math[ordinal_features]), columns=ordinal_features) #scaler_mm.transform(X_test[ordinal_features])
X_test_scaled_math = pd.concat([X_test_scaled_num_math.reset_index(drop=True),X_test_scaled_ord_math.reset_index(drop=True),X_test_rest_math.reset_index(drop=True)],axis=1)
```

```python
# scale the portuguese data
X_train_rest_port = X_train_port.loc[:, ~X_train_port.columns.isin(ordinal_features+numeric_features)]
X_test_rest_port = X_test_port.loc[:, ~X_test_port.columns.isin(ordinal_features+numeric_features)]

# fit and transform
scaler_mm.fit(X_train_port[ordinal_features])
scaler_std.fit(X_train_port[numeric_features])
X_train_scaled_num_port =  pd.DataFrame(scaler_std.transform(X_train_port[numeric_features]), columns=numeric_features)  #scaler_std.transform(X_train[numeric_features])
X_train_scaled_ord_port = pd.DataFrame(scaler_mm.transform(X_train_port[ordinal_features]), columns=ordinal_features)#scaler_mm.transform(X_train[ordinal_features])
X_train_scaled_port = pd.concat([X_train_scaled_num_port.reset_index(drop=True),X_train_scaled_ord_port.reset_index(drop=True),X_train_rest_port.reset_index(drop=True)],axis=1)


# transform the test data
X_test_scaled_num_port = pd.DataFrame(scaler_std.transform(X_test_port[numeric_features]), columns=numeric_features) #scaler_std.transform(X_test[numeric_features])
X_test_scaled_ord_port = pd.DataFrame(scaler_mm.transform(X_test_port[ordinal_features]), columns=ordinal_features) #scaler_mm.transform(X_test[ordinal_features])
X_test_scaled_port = pd.concat([X_test_scaled_num_port.reset_index(drop=True),X_test_scaled_ord_port.reset_index(drop=True),X_test_rest_port.reset_index(drop=True)],axis=1)
```

# Modeling
In this section, we will train 6 models:
1. Linear regression
2. Ridge regression
3. Lasso regression
4. Decision Tree
5. Random Forest
6. gradient boost regressor

For each model, the mean absolute error (MAE) and the root mean squared error (RMSE) are calculated. The scaled data will be used only for the linear regression, ridge and lasso models. 

```python
# Initialize the errors list
errors=[[],[]]
X_train_scaled = [X_train_scaled_math,X_train_scaled_port]
X_test_scaled = [X_test_scaled_math,X_test_scaled_port]
X_train=[X_train_math,X_train_port]
X_test=[X_test_math,X_test_port]
y_train=[y_train_math, y_train_port]
y_test=[y_test_math,y_test_port]
```

## Linear regression models 
### Linear Regression


```python
LR_model = LinearRegression()

for i in range(2):
    cv_results_LR = cross_validate(LR_model, X_train_scaled[i], y_train[i], cv=5, 
                            scoring=['neg_mean_absolute_error', 'neg_root_mean_squared_error'], 
                            return_train_score=True)
    LR_model.fit(X_train_scaled[i],y_train[i])

    y_te_pred = LR_model.predict(X_test_scaled[i])

    train_mae = -cv_results_LR['test_neg_mean_absolute_error'].mean()
    train_rmse = -cv_results_LR['test_neg_root_mean_squared_error'].mean()

    test_mae = mae(y_test[i], y_te_pred)
    test_rmse = np.sqrt(mse(y_test[i], y_te_pred))

    errors[i].append({'Model':'Linear regression',
               'MAE train': train_mae,
               'MAE test':test_mae,
               'RMSE train':train_rmse,
               'RMSE test':test_rmse})
    
```

```python
print("Math grades prediction\n", errors[0][0])
print("Portuguese grades prediction\n", errors[1][0])
```

### Ridge regression

```python
Ridge_model = Ridge()

for i in range(2):
    cv_results_Ridge = cross_validate(Ridge_model, X_train_scaled[i], y_train[i], cv=5, 
                            scoring=['neg_mean_absolute_error', 'neg_root_mean_squared_error'], 
                            return_train_score=True)
    Ridge_model.fit(X_train_scaled[i],y_train[i])

    y_te_pred = Ridge_model.predict(X_test_scaled[i])

    train_mae = -cv_results_Ridge['test_neg_mean_absolute_error'].mean()
    train_rmse = -cv_results_Ridge['test_neg_root_mean_squared_error'].mean()

    test_mae = mae(y_test[i], y_te_pred)
    test_rmse = np.sqrt(mse(y_test[i], y_te_pred))

    errors[i].append({'Model':'Ridge regression',
               'MAE train': train_mae,
               'MAE test':test_mae,
               'RMSE train':train_rmse,
               'RMSE test':test_rmse})
```

```python
print("Math grades prediction\n", errors[0][1])
print("Portuguese grades prediction\n", errors[1][1])
```

### Lasso regression

```python
Lasso_model = Lasso()
for i in range(2):
    cv_results_Lasso = cross_validate(Lasso_model, X_train_scaled[i], y_train[i], cv=5, 
                            scoring=['neg_mean_absolute_error', 'neg_root_mean_squared_error'], 
                            return_train_score=True)
    Lasso_model.fit(X_train_scaled[i],y_train[i])

    y_te_pred = Lasso_model.predict(X_test_scaled[i])

    train_mae = -cv_results_Lasso['test_neg_mean_absolute_error'].mean()
    train_rmse = -cv_results_Lasso['test_neg_root_mean_squared_error'].mean()

    test_mae = mae(y_test[i], y_te_pred)
    test_rmse = np.sqrt(mse(y_test[i], y_te_pred))

    errors[i].append({'Model':'Lasso regression',
               'MAE train': train_mae,
               'MAE test':test_mae,
               'RMSE train':train_rmse,
               'RMSE test':test_rmse})
```

```python
print("Math grades prediction\n", errors[0][2])
print("Portuguese grades prediction\n", errors[1][2])
```

## Tree based models 
### Decision Tree model


```python
DT_model = DecisionTreeRegressor(criterion="squared_error")

for i in range(2):
    cv_results_DT = cross_validate(DT_model, X_train[i], y_train[i], cv=5, 
                            scoring=['neg_mean_absolute_error', 'neg_root_mean_squared_error'], 
                            return_train_score=True)
    
    DT_model.fit(X_train[i],y_train[i])

    y_te_pred = DT_model.predict(X_test[i])

    train_mae = -cv_results_DT['test_neg_mean_absolute_error'].mean()
    train_rmse = -cv_results_DT['test_neg_root_mean_squared_error'].mean()

    test_mae = mae(y_test[i], y_te_pred)
    test_rmse = np.sqrt(mse(y_test[i], y_te_pred))

    errors[i].append({'Model':'Decision Tree',
               'MAE train': train_mae,
               'MAE test':test_mae,
               'RMSE train':train_rmse,
               'RMSE test':test_rmse})
```

```python
print("Math grades prediction\n", errors[0][3])
print("Portuguese grades prediction\n", errors[1][3])
```

### Random Forest model

```python
RF_model = RandomForestRegressor(random_state=47,criterion='squared_error')
for i in range(2):
    cv_results_RF = cross_validate(RF_model, X_train[i], y_train[i], cv=5, 
                            scoring=['neg_mean_absolute_error', 'neg_root_mean_squared_error'], 
                            return_train_score=True)
    RF_model.fit(X_train[i],y_train[i])

    y_te_pred = RF_model.predict(X_test[i])

    train_mae = -cv_results_RF['test_neg_mean_absolute_error'].mean()
    train_rmse = -cv_results_RF['test_neg_root_mean_squared_error'].mean()
    
    test_mae = mae(y_test[i], y_te_pred)
    test_rmse = np.sqrt(mse(y_test[i], y_te_pred))

    errors[i].append({'Model':'Random Forest',
               'MAE train': train_mae,
               'MAE test':test_mae,
               'RMSE train':train_rmse,
               'RMSE test':test_rmse})
```

```python
print("Math grades prediction\n", errors[0][4])
print("Portuguese grades prediction\n", errors[1][4])
```

### Gradient Boosting model


```python
GB_model = GradientBoostingRegressor(loss='huber',random_state=42)
for i in range(2):
    cv_results_GB = cross_validate(GB_model, X_train[i], y_train[i], cv=5, 
                            scoring=['neg_mean_absolute_error', 'neg_root_mean_squared_error'], 
                            return_train_score=True)
    GB_model.fit(X_train[i],y_train[i])

    #y_tr_pred = GB_model.predict(X_train[i])
    y_te_pred = GB_model.predict(X_test[i])

    train_mae = -cv_results_GB['test_neg_mean_absolute_error'].mean()
    train_rmse = -cv_results_GB['test_neg_root_mean_squared_error'].mean()

    test_mae = mae(y_test[i], y_te_pred)
    test_rmse = np.sqrt(mse(y_test[i], y_te_pred))

    errors[i].append({'Model':'Gradient Boosting',
               'MAE train': train_mae,
               'MAE test':test_mae,
               'RMSE train':train_rmse,
               'RMSE test':test_rmse})
```

```python
print("Math grades prediction\n", errors[0][5])
print("Portuguese grades prediction\n", errors[1][5])
```

# Hyperparameter Tuning
We will use Grid Search to fetch for the best parameters for each of the 3 models, fit the best estimators and assess their performance. 



```python
errors_courses = {'math':[],
                 'portuguese':[]}
courses = ['math','portuguese']

best_models = []
```

## Hyperparameter Tuning for Random Forest
### Fitting the model and assessing performance

```python
RF_model =  RandomForestRegressor(random_state=47) 
    
grid_params = {
        'n_estimators': [10, 30, 50, 80, 100, 130],
        'criterion' : ['absolute_error', 'squared_error'],
        'max_depth': [2,3, 5, 10, 20],
        #'max_features' : [None, 'sqrt', 'log2']
        
}
rf_grid_cv = GridSearchCV(RF_model, param_grid=grid_params, cv=5, scoring='neg_mean_squared_error', n_jobs=-1)
```

```python
for i in range(2):
    rf_grid_cv.fit(X_train[i], y_train[i])
    print("Best parameters for ",courses[i],": ",rf_grid_cv.best_params_)
    best_model_RF = rf_grid_cv.best_estimator_
    best_models.append(best_model_RF)
    rf_best_cv_results = cross_validate(best_model_RF, X_train[i], y_train[i], cv=5,scoring=['neg_mean_absolute_error', 'neg_root_mean_squared_error'])

    y_pred = best_model_RF.predict(X_test[i])

    rf_mae_mean_train = -rf_best_cv_results['test_neg_mean_absolute_error'].mean()
    rf_rmse_mean_train = -rf_best_cv_results['test_neg_root_mean_squared_error'].mean()

    rf_mae_std_train = rf_best_cv_results['test_neg_mean_absolute_error'].std()
    rf_rmse_std_train = rf_best_cv_results['test_neg_root_mean_squared_error'].std()

    rf_mae_test = mae(y_test[i],y_pred )
    rf_rmse_test =  np.sqrt(mse(y_test[i], y_pred)) 

    errors_courses[courses[i]].append({'Model':'Random Forest',
               'Mean MAE train': rf_mae_mean_train,
               'Std MAE train mean' : rf_mae_std_train,
               'MAE test':rf_mae_test,
               'Mean RMSE train':rf_rmse_mean_train,
               'Std RMSE train':rf_rmse_std_train,
               'RMSE test':rf_rmse_test})
```

### Feature importances of the best random forest regressor 
#### Math course

```python
plt.subplots(figsize=(10, 5))
imps = best_models[0].feature_importances_
rf_feat_imps = pd.Series(imps, index=X_train[0].columns).sort_values(ascending=False)
rf_feat_imps.plot(kind='bar')
plt.xlabel('features')
plt.ylabel('importance')
plt.title('Best random forest regressor feature importances (Math course)');
```

For math course, the 5 most important features are the number of absences, the grades' trend (difference between second and first year), the number of failures, free time, and student's health.

#### Portuguese course

```python
plt.subplots(figsize=(10, 5))
imps = best_models[1].feature_importances_
rf_feat_imps = pd.Series(imps, index=X_train[1].columns).sort_values(ascending=False)
rf_feat_imps.plot(kind='bar')
plt.xlabel('features')
plt.ylabel('importance')
plt.title('Best random forest regressor feature importances (Portuguese course)');
```

For the Portuguese course, there are 3 features in common with the mathematics course, among the 5 most important, namely grades' trend, number of failures and number of absences. Students' interest in higher education and the time spent with friends are also in the top 5 important features.

## Hyperparameter Tuning for Gradien Boosting
### Fitting the model and assessing performance

```python
GB_model =  GradientBoostingRegressor(random_state=42) 
    
grid_params = {
    'learning_rate': np.arange(0.05, 1, 0.05),
    'max_depth': np.arange(2, 10, 1),
    'n_estimators': np.arange(50, 200, 50),
    'loss':['squared_error','absolute_error','huber']
    #'max_features': [None, 'sqrt', 'log2']
    
}
GB_grid_cv = GridSearchCV(GB_model, param_grid=grid_params, cv=5, scoring='neg_mean_squared_error', n_jobs=-1)
```

```python
for i in range(2):
    GB_grid_cv.fit(X_train[i], y_train[i])
    print("Best parameters for ",courses[i],": ",GB_grid_cv.best_params_)
    best_model_GB = GB_grid_cv.best_estimator_
    best_models.append(best_model_GB)
    GB_best_cv_results = cross_validate(best_model_GB, X_train[i], y_train[i], cv=5,scoring=['neg_mean_absolute_error', 'neg_root_mean_squared_error'])

    y_pred = best_model_GB.predict(X_test[i])

    GB_mae_mean_train = -GB_best_cv_results['test_neg_mean_absolute_error'].mean()
    GB_rmse_mean_train = -GB_best_cv_results['test_neg_root_mean_squared_error'].mean()

    GB_mae_std_train = GB_best_cv_results['test_neg_mean_absolute_error'].std()
    GB_rmse_std_train = GB_best_cv_results['test_neg_root_mean_squared_error'].std()

    GB_mae_test = mae(y_test[i],y_pred )
    GB_rmse_test =  np.sqrt(mse(y_test[i], y_pred)) 

    errors_courses[courses[i]].append({'Model':'Gradient Boosting',
               'Mean MAE train': GB_mae_mean_train,
               'Std MAE train mean' : GB_mae_std_train,
               'MAE test':GB_mae_test,
               'Mean RMSE train':GB_rmse_mean_train,
               'Std RMSE train':GB_rmse_std_train,
               'RMSE test':GB_rmse_test})
```

### Feature importances of the best Gradient boosting regressor 
#### Math course

```python
plt.subplots(figsize=(10, 5))
imps = best_models[2].feature_importances_
GB_feat_imps = pd.Series(imps, index=X_train[0].columns).sort_values(ascending=False)
GB_feat_imps.plot(kind='bar')
plt.xlabel('features')
plt.ylabel('importance')
plt.title('Best Gradient boosting regressor feature importances (Math course)');
```

Gradient boosting identifies also the number of failures, absences and grades' trend as the top 3 most important features for the math course. 

#### Portuguese course

```python
plt.subplots(figsize=(10, 5))
imps = best_models[3].feature_importances_
GB_feat_imps = pd.Series(imps, index=X_train[1].columns).sort_values(ascending=False)
GB_feat_imps.plot(kind='bar')
plt.xlabel('features')
plt.ylabel('importance')
plt.title('Best Gradient boosting regressor feature importances (Portuguese course)');
```

For the portuguese course, in the 5 most important features for the Gradient Boosting, there are 3 common features with the Random Forest model which are *grade_trend*, *failures* and *absences*. This model emphasizes two other features namely *age* and *Dalc*.

# Conclusion

In order to predict the G3 grade, we have considered & regression models:

1. Linear regression
2. Ridge regression
3. Lasso regression
4. Decision Tree
5. Random Forest
6. gradient boost regressor
7. XGBoost regressor

   
After splitting the dataset into train and validation set, we scaled the data, this scaled data is used only for the linear, ridge and lasso regression. Tree based models are trained on non scaled data (these models are not distance based algorithms). 

We have evaluated each models in terms of mean absolute error (MAE) and root mean squared error (RMSE) for both the training and test data. The three best performing models are the Random forest, Gradient boost and Extreme gradient boost.

Next, we have applied the grid search CV for the hyperparameter tuning for the top three models seperately and we have assessed their performance. This step is very time consuming due to the use of grid search, this could be improved by the use of random search.

The three models have very close results, Random Forest slightly outperforms the other two models in terms of MAE wheras XGBOOST has a better RMSE.
