## Imports

```python
import pandas as pd
import numpy as np
import os
import pickle
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import scale
from sklearn.model_selection import train_test_split, cross_validate, GridSearchCV
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor,GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_squared_error as mse, mean_absolute_error as mae
from sklearn.pipeline import make_pipeline,Pipeline
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_regression
from xgboost import XGBRegressor
```

## Load The Data
We start by loading  the saved data from the previous step.

```python
df_student = pd.read_csv('./data/data_student_cleaned_pre.csv')
df_student.head()
```

Let's check the columns and their types. All features are numeric and there is no null value.

```python
df_student.info()
```

# Train test split

We split the data into a training and test sets:
- 30% for test
- 70% for training

```python
X = df_student.drop(columns=['G3'])
y = df_student['G3']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=47)
```

```python
print(X_train.shape)
print(X_test.shape)
```

# Modeling
In this section, we will train 7 models:
1. Linear regression
2. Ridge regression
3. Lasso regression
4. Decision Tree
5. Random Forest
6. gradient boost regressor
7. XGBoost regressor

for each model, we will tune its parameters to choose the best ones. A pipeline is created including a scaling step and the model, the scaling step is omitted for tree based models. The absolute mean error (MAE) and the root mean squared error (RMSE) are calculated for each parameter


## Linear Regression
### Pipeline creation and hyperparameter tuning 

```python
pipeline_lr = Pipeline(steps=[
    ('standardscaler', StandardScaler()),
    ('selectkbest', SelectKBest(score_func=f_regression)),
    ('linear_regression', LinearRegression())  
])

#pipe.get_params().keys()
k = [k+1 for k in range(len(X_train.columns))]
grid_params = {
    'standardscaler': [StandardScaler(), MinMaxScaler(), None],
    'selectkbest__k': k
    
}
lr_grid_cv = GridSearchCV(pipeline_lr, param_grid=grid_params, cv=5, n_jobs=-1)
lr_grid_cv.fit(X_train, y_train)
lr_grid_cv.best_params_
```

### Fit the best estimator

```python
best_model1 = pipeline_lr.set_params(**lr_grid_cv.best_params_)

best_model1.fit(X_train, y_train)
```

### Assess performance

```python
y_tr_pred = best_model1.predict(X_train)
y_te_pred = best_model1.predict(X_test)

print('r2_train = ',r2_score(y_train, y_tr_pred),'r2_test = ', r2_score(y_test, y_te_pred))
print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```

## Regularization
### Ridge regression

```python
pipeline_ridge = Pipeline(steps=[
    ('feature_selection', SelectKBest(score_func=f_regression)),  
    ('scaler', StandardScaler()),  
    ('ridge', Ridge(alpha=1.0))     
])
k = [k+1 for k in range(len(X_train.columns))]
param_grid = {
    'feature_selection__k': k,  # Number of features to select
    'ridge__alpha': [0.1, 1.0, 10.0],    # Regularization strength
    'scaler': [StandardScaler(), MinMaxScaler(), None],
}
ridge_grid_cv = GridSearchCV(pipeline_ridge, param_grid=param_grid, cv=5, n_jobs=-1)
ridge_grid_cv.fit(X_train, y_train)
ridge_grid_cv.best_params_
```

```python
best_model2 = pipeline_ridge.set_params(**ridge_grid_cv.best_params_)

best_model2.fit(X_train, y_train)
```

```python
y_tr_pred = best_model2.predict(X_train)
y_te_pred = best_model2.predict(X_test)

print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```

### Lasso regression

```python
pipeline_lasso = Pipeline(steps=[
    ('feature_selection', SelectKBest(score_func=f_regression)),  
    ('scaler', StandardScaler()),  
    ('lasso', Lasso(alpha=1.0))     
])
k = [k+1 for k in range(len(X_train.columns))]
param_grid = {
    'feature_selection__k': k,  # Number of features to select
    'lasso__alpha': [0.1, 1.0, 10.0],    # Regularization strength
    'scaler': [StandardScaler(), MinMaxScaler(), None],
}
lasso_grid_cv = GridSearchCV(pipeline_lasso, param_grid=param_grid, cv=5, n_jobs=-1)
lasso_grid_cv.fit(X_train, y_train)
lasso_grid_cv.best_params_
```

```python
best_model3 = pipeline_lasso.set_params(**lasso_grid_cv.best_params_)

best_model3.fit(X_train, y_train)
```

```python
y_tr_pred = best_model3.predict(X_train)
y_te_pred = best_model3.predict(X_test)

print('r2_train = ',r2_score(y_train, y_tr_pred),'r2_test = ', r2_score(y_test, y_te_pred))
print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```

## Tree based models 
### Decision Tree model
#### Pipeline creation and hyperparameter tuning

```python
pipeline_DT = Pipeline(steps=[
    ('scaler', StandardScaler()),  
    ('DT', DecisionTreeRegressor(criterion="squared_error"))     
])
k = [k+1 for k in range(len(X_train.columns))]
param_grid = {
    'scaler': [StandardScaler(), MinMaxScaler(), None],
    'DT__criterion' : ['absolute_error', 'squared_error'],
    'DT__max_depth': [3, 5, 10, 20, None],
    'DT__max_features' : [None, 'sqrt', 'log2']
    
}
DT_grid_cv = GridSearchCV(pipeline_DT, param_grid=param_grid, cv=5, n_jobs=-1)
DT_grid_cv.fit(X_train, y_train)
DT_grid_cv.best_params_
```

```python
best_model4 = pipeline_DT.set_params(**DT_grid_cv.best_params_)

best_model4.fit(X_train, y_train)
```

```python
y_tr_pred = best_model4.predict(X_train)
y_te_pred = best_model4.predict(X_test)

print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```

### Bagging model: Random Forest model
#### Pipeline creation and hyperparameter tuning

```python
pipeline_RF = Pipeline(steps=[
    ('scaler', StandardScaler()),  
    ('RF', RandomForestRegressor(random_state=47,criterion='squared_error')) 
    
])


grid_params = {
        'RF__n_estimators': [10, 30, 50, 80, 100, 130, 160],
        'RF__criterion' : ['absolute_error', 'squared_error'],
        'RF__max_depth': [3, 5, 10, 20, None],
        'RF__max_features' : [None, 'sqrt', 'log2'],
        'scaler': [StandardScaler(), MinMaxScaler(), None]
}
rf_grid_cv = GridSearchCV(pipeline_RF, param_grid=grid_params, cv=5, scoring='neg_mean_squared_error', n_jobs=-1)
rf_grid_cv.fit(X_train, y_train)
rf_grid_cv.best_params_
```

#### Fit the best estimator

```python
best_model5 = pipeline_RF.set_params(**rf_grid_cv.best_params_)

best_model5.fit(X_train, y_train)
```

#### Assess performance

```python
y_tr_pred = best_model5.predict(X_train)
y_te_pred = best_model5.predict(X_test)

print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```

### Boosting Models
#### Gradient Boost model
##### Pipeline creation and hyperparameter tuning

```python
pipeline_gb = Pipeline(steps=[
    ('scaler', StandardScaler()),  
    ('gb', GradientBoostingRegressor(random_state=42))
])
                           
grid_params = {
    'gb__learning_rate': np.arange(0.05, 1, 0.05),
    'gb__max_depth': np.arange(3, 10, 1),
    'gb__n_estimators': np.arange(50, 200, 50),
    'gb__max_features': [None, 'sqrt', 'log2'],
    'scaler': [StandardScaler(), MinMaxScaler(), None],
}
gb_grid_cv = GridSearchCV(pipeline_gb, param_grid=grid_params, cv=5, n_jobs=-1)
gb_grid_cv.fit(X_train, y_train)
gb_grid_cv.best_params_
```

##### Fit the best estimator

```python
best_model6 = pipeline_gb.set_params(**gb_grid_cv.best_params_)

best_model6.fit(X_train, y_train)
```

##### Assess performance

```python
y_tr_pred = best_model6.predict(X_train)
y_te_pred = best_model6.predict(X_test)

print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```

#### XGBoot model
##### Pipeline creation and hyperparameter tuning

```python
pipeline_xgb = Pipeline(steps=[
    ('scaler', StandardScaler()),  
    ('xgb', XGBRegressor(objective= 'reg:squarederror'))
])
                           
grid_params = {
    'xgb__learning_rate': np.arange(0.05, 1, 0.05),
    'xgb__max_depth': np.arange(3, 10, 1),
    'xgb__n_estimators': np.arange(50, 200, 50),
    'xgb__max_features': [None, 'sqrt', 'log2'],
    'xgb__alpha': [0, 0.1, 1, 10],
    'xgb__lambda': [0, 0.1, 1, 10]
    'xgb__scaler': [StandardScaler(), MinMaxScaler(), None],
}
xgb_grid_cv = GridSearchCV(pipeline_xgb, param_grid=grid_params, cv=5, n_jobs=-1)
xgb_grid_cv.fit(X_train, y_train)
xgb_grid_cv.best_params_
```

##### Fit the best estimator

```python
best_model7 = pipeline_xgb.set_params(**xgb_grid_cv.best_params_)

best_model7.fit(X_train, y_train)
```

##### Assess performance

```python
y_tr_pred = best_model7.predict(X_train)
y_te_pred = best_model7.predict(X_test)

print('mae_train = ',mae(y_train, y_tr_pred),'mae_test = ', mae(y_test, y_te_pred))
print('rmse_train = ',np.sqrt(mse(y_train, y_tr_pred)),'rmse_test = ', np.sqrt(mse(y_test, y_te_pred)))
```
